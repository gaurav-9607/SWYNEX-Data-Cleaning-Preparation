## 4. Extra Column Investigation

During the initial data quality inspection, an additional column named `Unnamed: 15` was identified.

The column contains 8 non-empty entries. The observed values include `0.0` and `32537.0`.

Since this column is not part of the original 15 attributes of the UCI Adult Census Income dataset, its associated records will be inspected before deciding whether the column should be removed.

```python
# Inspect all original attributes for the affected records

affected_records = df.loc[
    extra_rows.index,
    df.columns[:15]
].copy()

affected_records["Extra_Column_Value"] = df.loc[
    extra_rows.index, "Unnamed: 15"
]

display(affected_records)
```

```python
# Display the extra column values with their original row indexes

print("Extra column name:", df.columns[15])
print("\nAll non-empty values:")
print(df.loc[df["Unnamed: 15"].notna(), ["Unnamed: 15"]].to_string())

print("\nValue counts:")
print(df["Unnamed: 15"].value_counts(dropna=True))
```

## 5. Removal of Unnecessary Column

The investigation confirmed that the `Unnamed: 15` column is not one of the original 15 attributes of the UCI Adult Census Income dataset.

It contains 8 non-empty values: six entries of `32537.0` and two entries of `0.0`. These values do not represent a valid dataset attribute.

Therefore, only this unnecessary column will be removed. No data records will be deleted during this step.

```python
# Remove the unnecessary extra column

df = df.drop(columns=["Unnamed: 15"])

print("Extra column removed successfully.")
print("Updated dataset shape:", df.shape)
print("\nRemaining columns:")
print(df.columns.tolist())
```

## 8. Categorical Data Analysis

Categorical analysis helps understand how records are distributed across groups such as workclass, education, occupation, marital status, race, sex, native country, and income category.

Frequency counts are used to identify the most common categories in each variable.

```python
# Analyze the frequency distribution of categorical columns

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\nFrequency Distribution of {column}:")
    display(df[column].value_counts().head(10))
```

# 10. Data Visualization

Data visualization helps understand the distribution, patterns, and relationships within the dataset. In this section, numerical and categorical variables will be visualized using appropriate charts.

```python
# Visualize the distribution of numerical variables

import matplotlib.pyplot as plt

numerical_columns = df.select_dtypes(include="number").columns

plt.figure(figsize=(15, 10))

for i, column in enumerate(numerical_columns, 1):
    plt.subplot(2, 3, i)
    plt.hist(df[column], bins=30, edgecolor="black")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## Interpretation of Numerical Data Visualization

The histograms provide insights into the distribution of numerical variables:

1. **Age:** Most individuals fall within the working-age group.
2. **Fnlwgt:** The distribution is highly right-skewed, with a few large values.
3. **Education-num:** The values are discrete and represent different education levels.
4. **Capital-gain:** Most observations have zero capital gain.
5. **Capital-loss:** Most observations have zero capital loss.
6. **Hours-per-week:** A large number of individuals work approximately 40 hours per week.

Overall, the histograms help identify the distribution, concentration, and skewness of numerical variables.

```python
# Visualize the distribution of categorical variables

categorical_columns = df.select_dtypes(include=["object"]).columns.tolist()

print("Categorical Columns:", categorical_columns)

fig, axes = plt.subplots(3, 3, figsize=(18, 15))
axes = axes.flatten()

for i, column in enumerate(categorical_columns):
    df[column].value_counts().head(10).plot(
        kind="bar",
        ax=axes[i],
        edgecolor="black"
    )
    axes[i].set_title(f"Distribution of {column}")
    axes[i].set_xlabel(column)
    axes[i].set_ylabel("Frequency")
    axes[i].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

## Interpretation of Categorical Data Visualization

The bar charts illustrate the frequency distribution of categorical variables in the Adult Census Income dataset.

5. **Relationship:** Husband is the most common relationship category.
6. **Race:** White is the most frequently observed category.
